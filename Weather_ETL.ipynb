```python
cities = [
    "Cairo",
    "Giza",
    "Alexandria",
    "Port Said",
    "Suez",
    "Ismailia",
    "Mansoura",
    "Tanta",
    "Luxor",
    "Aswan"
]
```

```python
list_of_cities = []
```

```python
for city in cities:
    print (city)
    url = "https://api.openweathermap.org/data/2.5/weather"
    params = {
    "appid": "6a13e61e853763292101962a026aa3e8",
    "q": city
  }

    response = requests.get(url, params=params)
    data = response.json()
    city = city 
    country = "Egypt"
    lat = data['coord']['lat']
    lon = data['coord']['lon']
    temp = data['main']['temp']
    pressure = data['main']['pressure']
    Humidity = data['main']['humidity']
    wind_speed = data['wind']['speed']
    WeatherMain = data['weather'][0]['main']
    WeatherDescription = data['weather'][0]['description']

    list_of_cities.append({
        'city': city,
        'country': country,
        'lat': lat,
        'lon': lon,
        'temp': temp,
        'pressure': pressure,
        'humidity': Humidity,
        'wind_speed': wind_speed , 
        'WeatherMain':WeatherMain ,
        'WeatherDescription' : WeatherDescription
    })
```

```python
list_of_cities 
```

### 2. Transformation 

```python
import pandas as pd
```

```python
df = pd.DataFrame(list_of_cities)
df
```

```python
df.info()
```

```python
df['temp'] = df['temp']-273
df
```

```python
# Region Mapping
region_map = {
    "Cairo": "Greater Cairo",
    "Giza": "Greater Cairo",
    "Alexandria": "North Coast",
    "Port Said": "Canal",
    "Suez": "Canal",
    "Ismailia": "Canal",
    "Mansoura": "Lower Egypt",
    "Tanta": "Lower Egypt",
    "Luxor": "Upper Egypt",
    "Aswan": "Upper Egypt"
}

df["Region"] = df["city"].map(region_map)


# Temperature Category
def categorize_temperature(temp):
    if temp < 15:
        return "Cold"
    elif temp <= 25:
        return "Moderate"
    else:
        return "Hot"

df["TemperatureCategory"] = df["temp"].apply(categorize_temperature)


# Humidity Category
def categorize_humidity(humidity):
    if humidity < 40:
        return "Low"
    elif humidity <= 70:
        return "Medium"
    else:
        return "High"


df["HumidityCategory"] = df["humidity"].apply(categorize_humidity)
```
